### 1: Reading a TXT File
**Task Description:** You are provided with a text file (products.txt) that contains descriptions of various products. Your task is to read the file and display the content.

**Your Task:** Write a Python code snippet to:
1. Open the file in read mode.
2. Read the entire content of the file.
3. Print the content to the console.
4. Close the file.

In the cell below, write and execute your code.

```python
# Write and execute the code here
file = open("products.txt", "r")

content = file.read()

print(content)

file.close()
```

### 3: Reading a JSON File
**Task Description:** A JSON file (customers.json) contains customer information. Your task is to load the file and print the customer names.

**Your Task:** Write a Python code snippet to:
1. Import the json module.
2. Read the JSON file.
3. Extract and print the customer names.

In the cell below, write and execute your code.

```python
# Write and execute the code here
import json


with open("customers.json", "r") as file:
    customers = json.load(file)


for customer in customers["customers"]:
    print(customer["name"])
```

### 4: Reading an XML File
**Task Description:** An XML file (books.xml) contains book catalog data. Your task is to extract book titles from the file.

**Your Task:** Write a Python code snippet to:
1. Import the xml.etree.ElementTree module.
2. Parse the XML file.
3. Extract and print all book titles.

In the cell below, write and execute your code.

```python
# Write and execute the code here
import xml.etree.ElementTree as ET

tree = ET.parse("books.xml")
root = tree.getroot()

for book in root.findall(".//book"):
    title = book.find("title")
    if title is not None:
        print(title.text)
```
